Data Cleaning

```python
#copying 2021 results to a new df

results_2021_clean = results_2021_df[[
    'Year',
    'Constituency_Name',
    'Candidate',
    'Party',
    'Votes',
    'Vote_Share_Percentage'
]].copy()
```

```python
#choosing only the required columns

results_2021_clean.columns = [
    'election_year',
    'constituency_name',
    'candidate',
    'party',
    'total_votes',
    'vote_share'
]
```

```python
#copying 2026 results to a new df

results_2026_clean = results_2026_df[[
    'AC Name',
    'Candidate',
    'Party',
    'Total Votes',
    'Vote Share'
]].copy()
```

```python
results_2026_clean['election_year'] = 2026
```

```python
#choosing only the required columns
#Rearranging the columns

results_2026_clean = results_2026_clean[[
    'election_year',
    'AC Name',
    'Candidate',
    'Party',
    'Total Votes',
    'Vote Share'
]]
```

```python
results_2026_clean.columns = [
    'election_year',
    'constituency_name',
    'candidate',
    'party',
    'total_votes',
    'vote_share'
]
```

```python
#Combining 2021 and 2026 results for Dashboard purpose

combined_results = pd.concat(
    [results_2021_clean, results_2026_clean],
    ignore_index=True
)
```

```python
#removing missing values

combined_results = combined_results.dropna()
```

```python
#droping duplicates

combined_results = combined_results.drop_duplicates()
```
